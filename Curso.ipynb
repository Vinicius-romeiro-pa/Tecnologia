## If, elif e else

```python
tempo = 31

if tempo > 30:
    print("Está quente.")
    print("Bom dia")
    
```

```python
tempo = -10

if tempo >= 30:
    print("Está quente.")

elif tempo < 0:
   print("Está muito frio.")
```

```python
tempo = 5

if tempo >= 30:
    print("Está quente.")

elif tempo < 0:
   print("Está muito frio.")

else:
    print("A temperatura está normal.")
```

```python
Comando	  Significado	              Quando executa
if	         SE	           Se a primeira condição for verdadeira
elif	  SENÃO SE	       Se as anteriores forem falsas e sua condição for verdadeira
else	   SENÃO	           Se nenhuma condição anterior for verdadeira


if → Inicia a condição.

elif → Testa outra possibilidade.

else → Executa quando nenhuma condição anterior é atendida.

: → Indica o início do bloco de código.
```

## While

O while é um comando de repetição (loop) que executa um bloco de código enquanto uma condição for verdadeira (True).

```python
x = 0

while x < 10:
    print("0 valor de x é: {}".format(x))
    x = x + 1
```

```python
x = 0

while x < 10:
    print("0 valor de x é: {}".format(x))
    x += 1 # x = x + 1

else:
    print("concluído")
```

```python
x = 0

while x < 10:
    print("0 valor de x é: {}".format(x))
    print("x ainda é menor que 10, incrementando...")
    x += 1 # x = x + 1

else:
    print("concluído")
```

```python
x = 0

while x < 10:
    print("0 valor de x é: {}".format(x))
    print("x ainda é menor que 10, incrementando...")
    x += 1
    break

else:
    print("concluído")
```

```python
for variavel in iteravel:
    codigo
    codigo
```

```python
while true:
    codigo
```

```python
Explicação:
- while x < 5: → Repete enquanto x for menor que 5.
- print(x) → Mostra o valor de x.
- x += 1 → Aumenta x em 1 a cada repetição.
- Quando x chega a 5, o loop termina.



    Resumo para memorizar
Comando	         Significado
while	      Repete enquanto a condição for verdadeira
break	      Interrompe a repetição
continue	  Pula para a próxima repetição
x += 1	      Aumenta x em 1
x -= 1	      Diminui x em 1

while True:	Repete indefinidamente, até ocorrer uma interrupção

Diferença entre for e while
for	                                                                   while
Percorre uma sequência de elementos	                        Repete enquanto uma condição for verdadeira
Muito usado quando sabemos quantas repetições queremos	    Muito usado quando não sabemos quantas repetições serão necessárias
Exemplo: for i in range(5):	                                Exemplo: while x < 5:
```

```python
x=0
while x <= 10:
    print(x)
    x += 1
```

```python
y=10
while True: 
    print(y) 
    y -= 1 
    if y == 5: 
        print(y) 
        break
```
